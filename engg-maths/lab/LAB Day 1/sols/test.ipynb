```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
```

```python
# constants
c1 = 26.525
c2 = 0.0619
c3 = 3.343e-3
b  = 11.58e-5
g  = 9.807
```

```python
def system(t, y):
    z, v = y
    
    acc = (
        c1*np.exp(-b*z)
        - np.sign(v)*c2*(c3*np.abs(v)**0.5 + 0.5407*np.exp(-b*z/2)*np.abs(v)**2)
        - g
    )
    
    return [v, acc]
```

```python
t_span = (0, 3500)
t_eval = np.linspace(0, 3500, 2000)

y0 = [0, 0]

sol = solve_ivp(system, t_span, y0, t_eval=t_eval)
```

```python
t = sol.t
z = sol.y[0]
v = sol.y[1]
```

```python
plt.figure()
plt.plot(t, z)
plt.xlabel("t")
plt.ylabel("z")
plt.title("z vs t")
plt.grid()
plt.show()
```

```python
plt.figure()
plt.plot(t, v)
plt.xlabel("t")
plt.ylabel("dz/dt")
plt.title("dz/dt vs t")
plt.grid()
plt.show()
```

```python
plt.figure()
plt.plot(z, v)
plt.xlabel("z")
plt.ylabel("dz/dt")
plt.title("dz/dt vs z")
plt.grid()
plt.show()
```
